```python
from pathlib import Path

RAW = Path("../data/raw/wikipedia")

for p in sorted(RAW.glob("pl_*.html")):
    html = p.read_text(encoding="utf-8")
    i = html.find("departure")
    if i == -1:
        print(f"{p.name}: the word 'departure' is NOT in the page")
    else:
        print(f"{p.name}: ...{html[i - 60 : i + 15]!r}")
```
