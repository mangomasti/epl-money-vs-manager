```python
import json
from pathlib import Path

RAW = Path("../data/raw/understat")
data = json.loads((RAW / "epl_2014.json").read_text(encoding="utf-8"))

print("Top-level keys:", list(data.keys()))
print("\nFirst match record:")
print(json.dumps(data["dates"][0], indent=2))
```

```python
fd = pd.read_csv("../data/processed/matches.csv")
fd_teams = set(fd["home_team"])

us_teams = set()
for f in sorted(RAW.glob("epl_*.json")):
    season = json.loads(f.read_text(encoding="utf-8"))
    us_teams |= {m["h"]["title"] for m in season["dates"]}

print(f"football-data: {len(fd_teams)} teams | Understat: {len(us_teams)} teams\n")
print("Same name in both:     ", len(fd_teams & us_teams))
print("Only in football-data: ", sorted(fd_teams - us_teams))
print("Only in Understat:     ", sorted(us_teams - fd_teams))
```

```python
for name in sorted(us_teams - fd_teams):
    print(name)
```

```python
fd = pd.read_csv("../data/processed/matches.csv", parse_dates=["date"])
us = pd.read_csv("../data/processed/understat_matches.csv", parse_dates=["date"])
m = fd.merge(us, on=["season", "home_team", "away_team"], suffixes=("", "_us"))

diff = m[m["date"] != m["date_us"]].copy()
diff["days_off"] = (diff["date_us"] - diff["date"]).dt.days

# Bring back Understat's original kickoff time from the raw files
raw_times = {}
for f in RAW.glob("epl_*.json"):
    for match in json.loads(f.read_text(encoding="utf-8"))["dates"]:
        raw_times[int(match["id"])] = match["datetime"]
diff["understat_kickoff"] = diff["understat_id"].map(raw_times)

print("Disagreements by season and size:")
print(diff.groupby(["season", "days_off"]).size().to_string())
diff[["season", "date", "home_team", "away_team", "understat_kickoff", "days_off"]]
```
